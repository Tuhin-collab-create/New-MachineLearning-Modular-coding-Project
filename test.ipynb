```python
import numpy as np

# 1. Define the class or just the method inside your pipeline/class
class DeliveryPreprocessor:
    def distance_numpy(self, df, lat1, lon1, lat2, lon2):
        p = np.pi / 180
        a = (0.5 - np.cos((df[lat2] - df[lat1]) * p) / 2 + 
             np.cos(df[lat1] * p) * np.cos(df[lat2] * p) * 
             (1 - np.cos((df[lon2] - df[lon1]) * p)) / 2)
        
        df['distance'] = 12734 * np.arcsin(np.sqrt(np.clip(a, 0.0, 1.0)))
        return df

# 2. Instantiate your class
processor = DeliveryPreprocessor()

# 3. Apply it to your dataframe using your column names
df = processor.distance_numpy(
    df=df, 
    lat1='Restaurant_latitude', 
    lon1='Restaurant_longitude', 
    lat2='Delivery_location_latitude', 
    lon2='Delivery_location_longitude'
)

# Preview the result
print(df[['Restaurant_latitude', 'Restaurant_longitude', 'Delivery_location_latitude', 'Delivery_location_longitude', 'distance']].head())
```
